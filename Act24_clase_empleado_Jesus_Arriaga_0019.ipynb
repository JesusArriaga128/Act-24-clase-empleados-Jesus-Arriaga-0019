```python
### Jesus Arriaga NL = 5 | 2do H | Fecha: 05/Mayo/2026

class Empleado:
    def __init__(self, nombre, cargo, salario, fechaIngreso):
        self.nombre = nombre
        self.cargo = cargo
        self.salario = salario
        self.fechaIngreso = fechaIngreso

    def leerdatos(self):
        self.nombre = input("Nombre del empleado: ")
        self.cargo = input("Cargo del empleado: ")
        self.salario = input("Salario mensual: ")
        self.fechaIngreso = input("Fecha de ingreso: ")

    def registrarHoras(self):
        # Función interactiva para registrar horas
        horas = input(f"Horas trabajadas por {self.nombre}: ")
        print(f"-> Sistema: Se han registrado {horas} horas.")

    def solicitarVacaciones(self):
        # Función interactiva para solicitar días
        dias = input(f"Días de vacaciones solicitados por {self.nombre}: ")
        print(f"-> Sistema: Solicitud por {dias} días procesada.")

    def obtener_formato(self):
        resumen = f"Empleado: {self.nombre} | Cargo: {self.cargo} | Salario: ${self.salario} | Ingreso: {self.fechaIngreso}"
        return resumen

    def generarReporte(self):
        # Equivale a mostrardatos()
        print(self.obtener_formato())

# --- Creación y uso de objetos ---

print("Datos del Empleado 1")
empleado1 = Empleado("", "", "", "")
empleado1.leerdatos()
empleado1.registrarHoras()
empleado1.solicitarVacaciones()
empleado1.generarReporte()

print("*_*_*_*_*_*_*_*_*_*_*__*_*_* ")

print("Datos del Empleado 2")
empleado2 = Empleado("", "", "", "")
empleado2.leerdatos()
empleado2.registrarHoras()
empleado2.solicitarVacaciones()
empleado2.generarReporte()

print("\nPrograma adaptado por Jesus Arriaga NL = 5")
```
